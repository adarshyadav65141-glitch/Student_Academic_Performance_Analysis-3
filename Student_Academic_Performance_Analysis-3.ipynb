```python

# 10. Create a line graph of semester-wise average marks
import matplotlib.pyplot as plt
avg_marks = data.groupby("Semester")["Marks"].mean()

avg_marks.plot(kind="line", marker="o")

plt.xlabel("Semester")
plt.ylabel("Average Marks")
plt.title("Semester-wise Average Marks")
plt.show()
```

```python
# 11. Create a bar graph of subject-wise average marks

avg_marks = data.groupby("Subject")["Marks"].mean()

avg_marks.plot(kind="bar")

plt.xlabel("Subject")
plt.ylabel("Average Marks")
plt.title("Subject-wise Average Marks")

plt.xticks(rotation=90)

# plt.tight_layout()
plt.show()
```

```python
# 12. Identify your highest-performing subject.
highest_marks = max(
    sem3["Marks"].max(),
    sem4["Marks"].max(),
    sem5["Marks"].max(),
    sem6["Marks"].max()
)

print("Highest Marks:", highest_marks)
```

```python
# 13. Identify your lowest-performing subject.
Lowet_marks = min(
    sem3["Marks"].min(),
    sem4["Marks"].min(),
    sem5["Marks"].min(),
    sem6["Marks"].min()
)

print("Highest Marks:", Lowet_marks)
```

```python
# 14. Identify your best and worst semester?
semester_totals = {
    "Semester 3": sem3["Marks"].sum(),
    "Semester 4": sem4["Marks"].sum(),
    "Semester 5": sem5["Marks"].sum(),
    "Semester 6": sem6["Marks"].sum()
}

best_semester = max(semester_totals, key=semester_totals.get)
worst_semester = min(semester_totals, key=semester_totals.get)

print("Best Semester:", best_semester)
print("Worst Semester:", worst_semester)
```

```python
# Q16. NumPy से Mean, Median, Maximum, Minimum, Standard Deviation
import numpy as np

marks = data["Marks"].values

print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Maximum:", np.max(marks))
print("Minimum:", np.min(marks))
print("Standard Deviation:", np.std(marks))
```

```python
# 17. Create a graph showing the performance of each subject across the six semesters.
import matplotlib.pyplot as plt

subject_avg = data.groupby("Semester")["Marks"].mean()

subject_avg.plot(kind="line", marker="o")

plt.title("Subject Performance Across Six Semesters")
plt.xlabel("Semester")
plt.ylabel("Average Marks")
plt.legend(title="Subjects")
plt.grid()
plt.show()
```

```python
# 18. Set an academic target (target = 75%) and show it on your graph using axhline().
import matplotlib.pyplot as plt

subject_avg = data.groupby("Semester").mean(numeric_only=True)

ax = subject_avg.plot(kind="line", marker="o", figsize=(12, 6))

ax.axhline(y=75, color="red", linestyle="--", label="Target = 75%")

plt.title("Subject Performance Across Six Semesters")
plt.xlabel("Semester")
plt.ylabel("Average Marks")
plt.legend()
plt.grid()
plt.show()
```

```python
# 20. Create a 2×2 Matplotlib dashboard.
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Semester-wise average marks
semester_avg = data.groupby("Semester")["Marks"].mean()
axes[0, 0].plot(semester_avg.index, semester_avg.values, marker="o")
axes[0, 0].set_title("Average Marks by Semester")
axes[0, 0].set_xlabel("Semester")
axes[0, 0].set_ylabel("Average Marks")
axes[0, 0].grid()

# 2. Subject performance
subject_avg = data.groupby("Semester").mean(numeric_only=True)
subject_avg.plot(ax=axes[0, 1])
axes[0, 1].set_title("Subject Performance")

# 3. Marks distribution
axes[1, 0].hist(data["Marks"], bins=10)
axes[1, 0].set_title("Marks Distribution")
axes[1, 0].set_xlabel("Marks")
axes[1, 0].set_ylabel("Frequency")

# 4. Semester-wise marks
axes[1, 1].bar(semester_avg.index, semester_avg.values)
axes[1, 1].set_title("Semester Average Comparison")
axes[1, 1].set_xlabel("Semester")
axes[1, 1].set_ylabel("Average Marks")

plt.tight_layout()
plt.show()
```
